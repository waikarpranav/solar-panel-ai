# SolarPanelAI - Data Preprocessing
## 1. Introduction
Cleaning and preparing data for machine learning. We will handle time-series formats and scale numerical features.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
import joblib

df = pd.read_csv('../DATA/solar_weather_data.csv')
print("✓ Data loaded for preprocessing.")
```

## 2. Handling Dates and Times

```python
df['datetime'] = pd.to_datetime(df['datetime'])
df['month'] = df['datetime'].dt.month
df.drop(columns=['datetime'], inplace=True)
df.head()
```
